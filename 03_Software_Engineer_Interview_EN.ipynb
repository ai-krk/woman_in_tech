## 1. Meet the candidate and the role

**Role:** Junior Software Engineer, a fictional practice role.

The role involves explaining basic **Python or Java**, writing simple tests, investigating bugs,
using Git and discussing a personal contribution to a shared project. HTTP/API and SQL basics are
learning areas. Both languages, production experience and cloud deployment are not requirements here.
This is a teaching scenario, **not an HSBC job advertisement or actual hiring criteria**.

**Candidate:** Maya has course-project experience. There are three evidence records from two projects:

| Evidence ID | What the supplied record actually says |
|---|---|
| `PY_TASKS` | A local Python task-list program with add/show/remove functions. Maya handled blank titles and manually checked inputs. No database, external API or automated tests. |
| `JAVA_CATALOGUE` | A library catalogue built with a classmate. Maya wrote the title-search method and two JUnit tests: exact match and no result. It was shared work. |
| `DEBUG_SEARCH` | In that same Java project, Maya reproduced a letter-case search bug, agreed the expected behaviour, changed the comparison and added a regression test. A teammate reviewed her Git-branch change. |

There is no evidence of a commercial deployment, user numbers, performance improvement,
team leadership or professional employment. A missing record means **not established by these records**,
not proof that a real person lacks the ability.

### Three useful words

- **Unit test:** a small automated check of a specific part of a program.
- **Regression test:** a check added to help catch a previously fixed bug if it returns.
- **API:** a defined way for programs to communicate. Calling Gemini is one example;
  that does not mean Maya built an API in her course projects.

**Before asking the agent:** choose one thing you want help with. A focused request is easier to check
than "prepare me for every possible software engineering interview".

### S06. Let the agent retrieve facts

`get_evidence("all")` returns all three records for overview and planning prompts.
`all` is a lookup option, not a job skill. A lookup for SQL returns `missing`, not invented experience.


```python
def get_role(role_id):
    if not isinstance(role_id, str) or role_id not in ROLES:
        return {"status": "unsupported", "allowed_roles": list(ROLES)}
    return {"status": "found", "role_id": role_id, **ROLES[role_id]}

def get_evidence(skill):
    if not isinstance(skill, str) or skill not in ROLES[ROLE_ID]["skills"] + ["all"]:
        return {"status": "unsupported", "records": []}
    records = EVIDENCE if skill == "all" else [r for r in EVIDENCE if skill in r["skills"]]
    return {"status": "found" if records else "missing", "records": records}

pprint(get_evidence("sql"))
```

### S08. Keep the same three read-only tool boundaries

The tools can read the role, evidence and question bank. They cannot send an application, email anyone,
browse job sites or run generated Java/Python code. The local assertion checks a rejected action, not universal model safety.


```python
def get_question(skill, tone):
    if not isinstance(skill, str) or skill not in QUESTIONS:
        return {"status": "unsupported", "reason": "Unknown skill."}
    if not isinstance(tone, str) or tone not in TONES:
        return {"status": "unsupported", "reason": "Unknown tone."}
    return {"status": "found", "question": TONES[tone] + QUESTIONS[skill]}
```

```python
TOOL_RULES = {
    "get_role": {"function": get_role, "description": "Read the fictional junior software engineer role.",
                 "arguments": {"role_id": list(ROLES)}},
    "get_evidence": {"function": get_evidence,
                     "description": "Read fictional course evidence by skill. Use all for the complete three-record profile.",
                     "arguments": {"skill": list(QUESTIONS) + ["all"]}},
    "get_question": {"function": get_question, "description": "Read one authored beginner interview question.",
                     "arguments": {"skill": list(QUESTIONS), "tone": list(TONES)}},
}
TOOLS = describe_tools(TOOL_RULES)

def execute_tool(name, arguments):
    return execute_safe(name, arguments, TOOL_RULES)

assert execute_tool("send_email", {})["status"] == "rejected"
```

### S11. Use the same bounded agent loop

This is the main workshop's loop. Defining it sends no request.
During a live run, Gemini chooses tool calls, Python validates them and returns the results.
Each run allows at most five model requests and six tool attempts.


```python
def run_coach(request, active_client):
    session = AgentSession(request)
    for _ in range(session.max_requests):
        reply = session.ask(active_client, MODEL, INSTRUCTIONS, TOOLS)
        if reply is None:
            break
        calls = session.record(reply)
        if not calls:
            session.finish(getattr(reply, "output_text", ""))
            break
        if not session.can_execute(calls):
            break
        for call in calls:
            result = execute_tool(call.name, call.arguments)
            session.add_tool_result(call, result)
    return session.result()
```

### S12. Prepare one complete request at a time

Each scenario supplies a full prompt and selects a recipe. This helper replaces the previous selection;
it does not append old prompts or give the model an automatic chat history.


```python
def set_case(title, focus, recipe, prompt):
    global CURRENT_CASE, INSTRUCTIONS, REQUEST
    if focus not in QUESTIONS or recipe not in SKILLS:
        raise ValueError("Choose a supplied focus and recipe.")
    CURRENT_CASE = {"title": title, "focus": focus, "recipe": recipe}
    INSTRUCTIONS = RULES + "\n" + SKILLS[recipe]
    REQUEST = f"Target role ID: {ROLE_ID}. Focus for this exercise: {focus}.\n" + prompt
    print("Selected:", title)
    print("Recipe:", recipe)
    print("\nPrompt preview (not sent):\n" + REQUEST)
```
